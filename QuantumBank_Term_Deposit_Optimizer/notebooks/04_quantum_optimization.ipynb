# 04. Quantum Optimization (QAOA) vs Classical Benchmark
## QuantumBank Term Deposit Campaign Pipeline

```python
import sys
sys.path.append('..')
import pandas as pd
from src.classical_optimizer import ClassicalBruteForceOptimizer
from src.quantum_optimizer import QAOAOptimizer

top10 = pd.read_csv('../outputs/stage2_optimization_results.csv')

# Classical Brute Force
cl_opt = ClassicalBruteForceOptimizer(budget_limit=25.0, capacity_limit=22.0, k_select=6)
cl_sol, all_combos = cl_opt.solve(top10)
print('Classical Optimum:', cl_sol)

# QAOA Quantum Optimizer
qaoa_opt = QAOAOptimizer(p_cardinality=40.0, p_budget=15.0, p_capacity=15.0, shots=10000)
qaoa_sol, measured_df, qaoa_meta = qaoa_opt.solve(top10)
print('QAOA Optimum:', qaoa_sol)
print('Optimality Gap:', (cl_sol['expected_net_value'] - qaoa_sol['expected_net_value'])/cl_sol['expected_net_value'] * 100, '%')
```
